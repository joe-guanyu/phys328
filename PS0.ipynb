```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, norm
from ipywidgets import interact, IntSlider, FloatSlider

@interact(
    N=IntSlider(value=100, min=1, max=1000, continuous_update=True),
    p=FloatSlider(value=0.5, min=0.01, max=0.99, step=0.01,
                  continuous_update=True)
)
def plot(N, p):
    n = np.arange(N + 1)
    mu, sigma = N * p, np.sqrt(N * p * (1 - p))
    x = np.linspace(max(-0.5, mu - 4*sigma),
                    min(N + 0.5, mu + 4*sigma), 500)

    plt.figure(figsize=(8, 4))
    plt.bar(n, binom.pmf(n, N, p), alpha=0.5, label="Binomial")
    plt.plot(x, norm.pdf(x, mu, sigma), "r-", label="Normal")
    plt.xlim(x[0], x[-1])
    plt.xlabel("Number of successes")
    plt.ylabel("Probability / density (unit bins)")
    plt.title(f"Np={mu:.2f}, N(1-p)={N*(1-p):.2f}")
    plt.legend()
    plt.show()
```

## Binomial distribution approaching a normal distribution

The binomial distribution is

$$
P(n)=\binom{N}{n}p^n(1-p)^{N-n},
\qquad n=0,1,\ldots,N.
$$

The mean and variance are

$$
\mu=Np,\qquad \sigma^2=Np(1-p).
$$

The corresponding normal probability density is

$$
f(x)=\frac{1}{\sqrt{2\pi\sigma^2}}
\exp\left[-\frac{(x-\mu)^2}{2\sigma^2}\right].
$$

As $N$ increases at fixed $0<p<1$, the binomial distribution approaches to a normal distribution due to the central limit theorem.

$$
\frac{n-Np}{\sqrt{Np(1-p)}}
\;\xrightarrow{d}\;\mathcal{N}(0,1).
$$

### Why does this matter for statistical mechanics?

As $N \rightarrow \infty$, macroscopic quantities become increasingly predictable even though individual particles behave randomly.

The relative flutuation is

$$
\frac{\sigma}{\mu}
=\sqrt{\frac{1-p}{Np}}
\propto \frac{1}{\sqrt{N}}
\qquad \text{at fixed }p.
$$

It decreases and the macroscopic quantities become stable.
