```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson, norm

n, p = 1000, 0.03
lam = n * p
mu = n * p
sigma = np.sqrt(n * p * (1 - p))

k = np.arange(0, 120)
x = np.linspace(-0.5, 11.5, 300)

plt.figure(figsize=(8, 4.5))
plt.bar(k - 0.15, binom.pmf(k, n, p), width=0.3, label="Binomial(100, 0.03)", alpha=0.8)
plt.bar(k + 0.15, poisson.pmf(k, lam), width=0.3, label="Poisson(λ=3)", alpha=0.8)
plt.plot(x, norm.pdf(x, mu, sigma), 'k--', label="Normal(μ=3, σ=1.71)", linewidth=2)

plt.xlabel("k (Number of successes)")
plt.ylabel("Probability / Density")
plt.title("Binomial vs. Poisson vs. Normal Approximation")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
